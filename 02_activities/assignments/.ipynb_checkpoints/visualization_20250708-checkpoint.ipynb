```python
import PIL
import requests
import numpy as np
from matplotlib import pyplot as plt
import pandas as pd
import scipy
```

```python
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0,100,50)
y2 = np.random.randint(0,100,50)
```

```python
fig,ax = plt.subplots()
ax.plot(x,y1,label = 'Person 1')
ax.plot(x,y2,label = 'Person 2')
ax.legend(loc = 'lower right',
          frameon = True,
          fontsize = 12,
          ncol = 2,
          shadow = True,
          bbox_to_anchor = (1,1)
         )
fig.savefig('test.png')
```

```python
fig,ax = plt.subplots()
ax.plot(x,y1,label = 'Person 1')
ax.plot(x,y2,label = 'Person 2')
ax.legend(loc = 'upper left',
          frameon = True,
          fontsize = 12,
          ncol = 1,
          shadow = True,
          bbox_to_anchor = (1,1)
         )
fig.savefig('test.png')
```

```python
fig,ax = plt.subplots()
ax.scatter(x,y1,label = 'Person 1')
ax.scatter(x,y2,label = 'Person 2')
ax.legend(loc = 'upper left',
          frameon = True,
          fontsize = 12,
          ncol = 1,
          shadow = True,
          bbox_to_anchor = (1,1)
         )
ax.text(10,95,"This value is important!",
        ha = 'center',
        va = 'center',
        color = 'red',
        size = 20
       )
fig.savefig('test.png')
```

```python
fig,ax = plt.subplots()
ax.axis([0,10,0,10])
ax.text(1,5,". Data: (1,5)",transform = ax.transData)
ax.text(0.5,0.1,". Axes: (0.5,0.1)",transform = ax.transAxes)
ax.text(0.2,0.2,". Figure: (0.2, 0.2)",transform = fig.transFigure)
```

```python
np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0,75,50)
x2 = np.array(["Luffy","Zoro","Nami","Usopp","Sanji"])
y2 = np.array([110,180,240,99,220])

fig,(ax1,ax2) = plt.subplots(ncols = 2, 
                             nrows = 1,
                             figsize=(7,3)
                            )
ax1.scatter(x1,y1)
ax2.bar(x2,y2)
```

```python
x1 = np.arange(50)
y1 = np.random.randint(0,75,50)
x2 = np.array(["Luffy","Zoro","Nami","Usopp","Sanji"])
y2 = np.array([110,180,240,99,220])

fig,(ax1,ax2) = plt.subplots(ncols = 2, 
                             nrows = 1,
                             figsize=(7,3)
                            )
ax1.scatter(x1,y1)
ax2.bar(x2,y2,color = ["red","blue","green","magenta","indigo"])
ax1.set_xlabel("Time (minutes)")
ax1.set_ylabel("Count")
ax1.annotate("This value is important!",xy = (x1[10],y1[10]), xytext = (x1[10]+2,y1[10]+2),arrowprops = dict(arrowstyle = 'fancy'),rotation=40,color= "red")
ax2.set_ylabel("Count")
plt.tight_layout()
plt.savefig("test.png")
```

```python
fig,someaxes = plt.subplot_mosaic([['ax1','ax3'],
                                  ['ax2','ax3']],figsize=(7,4),layout = "constrained")
someaxes['ax1'].plot(x1,y1)
someaxes['ax2'].scatter(x1,y1)
someaxes['ax3'].bar(x2,y2)

someaxes['ax1'].set_xlabel("A Big Lable", fontsize = 18)
someaxes['ax2'].set_xlabel("Another label", fontsize = 18)
someaxes['ax3'].set_xlabel("Another one",fontsize = 18)
plt.show()
```
