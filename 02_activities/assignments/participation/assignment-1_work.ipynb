```python
# import all the necessary libraries
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
import os
```

```python
# provide a random sed to make the random numbers same - makes results reproducible
np.random.seed(613)
x = np.arange(50)
# generate 50 random integers from 0 to 100 (not including)
y = np.random.randint(0,100,50)
# create figure and axeses
fig,ax = plt.subplots(figsize=(5,3))
# ax.scatter(x,y)
# ax.bar(x,y)
# ax.plot(x,y)
ax.hist(y)

```

```python
# create dictionaries for stroring font parameters
font_label = {"family":"Arial","color":"blue","size":12} 
font_title = {"family":"Sans-serif","color":"red","size":16}
# create plot figure and axes handles
fig,ax = plt.subplots(figsize=(5,3))
ax.plot(x,y)
ax.set_title("Growth Chart",fontdict=font_title)
ax.set_xlabel("Days",fontdict = font_label)
ax.set_ylabel("Growth")
```

```python
# make another plot to show scatter plot
fig,ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y,marker="*",color="indigo",linestyle="-",linewidth=2)
```

```python
# Make scatter plot with line and playaround with line parameters
fig,ax = plt.subplots(figsize=(5,3))
ax.plot(x,y,marker="*",
           color="indigo",
           linestyle="--",
           linewidth=2,
           markersize=10
           )
```

```python
# About to get more crazy with the previous figure - using more plot config options!
fig,ax = plt.subplots(figsize=(5,3))
ax.plot(x,y,marker="*",
           color="#eeaa12",
           linestyle="--",
           linewidth=2,
           markersize=10,
           markerfacecolor="#555555",
           markerfacecoloralt="red",
           markeredgecolor="#123456"
           )
ax.grid(axis='x',color="blue",linewidth=2)
ax.grid(axis='y',color="red")
```
